```python
import pandas as pd

def augment_rows(row):
    new_rows = []

    #1
    new_rows.append(row)


    #2
    swapped_neg = row.copy()
    swapped_neg["negative_example_1"], swapped_neg["negative_example_2"] = \
        row["negative_example_2"], row["negative_example_1"]
    new_rows.append(swapped_neg)


    # 3.
    swapped_pos = row.copy()
    swapped_pos["positive_example_1"], swapped_pos["positive_example_2"] = \
        row["positive_example_2"], row["positive_example_1"]
    new_rows.append(swapped_pos)



    # 4.
    swapped_pos = row.copy()
    swapped_pos["positive_example_1"], swapped_pos["positive_example_2"] = row["positive_example_2"], row["positive_example_1"]
    swapped_pos["negative_example_1"], swapped_pos["negative_example_2"] = row["negative_example_2"], row["negative_example_1"]
    new_rows.append(swapped_pos)
    

    # +2. Rule-based body <-> examples swap
    if row["rule_violation"] == 1:
        for col in ["negative_example_1", "negative_example_2"]:
            swapped = row.copy()
            swapped["body"], swapped[col] = row[col], row["body"]
            new_rows.append(swapped)
    else:
        for col in ["positive_example_1", "positive_example_2"]:
            swapped = row.copy()
            swapped["body"], swapped[col] = row[col], row["body"]
            new_rows.append(swapped)

    return new_rows

# Apply to all rows
augmented = []
for _, row in df_train.iterrows():
    augmented.extend(augment_rows(row))

augmented_df = pd.DataFrame(augmented)

augmented_df.shape
```

```python
def make_context(row):
    return f"""
*discussions on this topic*:{row['subreddit']}
*rules of discussion*:{row['rule']}
*positive_examples(0)*:
1. {row['positive_example_1']}
2. {row['positive_example_2']}
*negative_examples(1)*:
1. {row['negative_example_1']}
2. {row['negative_example_2']}
*Direction for task*:
classify next statement accordingly.
"""

# Build new dataframe
new_df = pd.DataFrame({
    "context": augmented_df.apply(make_context, axis=1),
    "statement": augmented_df["body"],
    "target": augmented_df["rule_violation"]
})
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

# Shuffle dataframe
df_aug_context_statement_target = df_aug_context_statement_target.sample(frac=1, random_state=42).reset_index(drop=True)

# Create numpy arrays
X = df_aug_context_statement_target[["context", "statement"]].values   # shape (n_samples, 2)
Y = df_aug_context_statement_target["target"].values                   # shape (n_samples,)

# Train/test split (90:10) with stratification
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.1, random_state=42, stratify=Y
)

print("Train size:", X_train.shape, "Test size:", X_test.shape)
print("Class distribution in Y_train:", Y_train.shape)
print("Class distribution in Y_test:", Y_test.shape)
```

```python
import tensorflow as tf
from transformers import DistilBertTokenizerFast, TFDistilBertForSequenceClassification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import pandas as pd
import numpy as np
tokenizer = DistilBertTokenizerFast.from_pretrained("distilbert-base-uncased")
```

```python
all_lengths = df_aug_context_statement_target.apply(lambda row: len(tokenizer(row['context'], row['statement'])['input_ids']), axis=1)
print("Max length:", all_lengths.max())
print("Average length:", all_lengths.mean())
```

```python

MAX_LEN = 512

def tokenize(sent1, sent2, labels):
    encodings = tokenizer(
        sent1.tolist(),  # convert numpy array to list
        sent2.tolist(),
        truncation=True,
        padding="max_length",
        max_length=MAX_LEN,
        return_tensors="tf"
    )
    return encodings["input_ids"], encodings["attention_mask"], tf.convert_to_tensor(labels, dtype=tf.int32)

# If X_train and X_test are numpy arrays of shape (num_examples, 2)
train_input_ids, train_attention, train_labels = tokenize(
    X_train[:, 0],  # statement1
    X_train[:, 1],  # statement2
    Y_train         # labels
)

val_input_ids, val_attention, val_labels = tokenize(
    X_test[:, 0],  # statement1
    X_test[:, 1],  # statement2
    Y_test         # labels
)

# =====================================
# Build tf.data.Dataset
# =====================================
train_ds = tf.data.Dataset.from_tensor_slices(({
        "input_ids": train_input_ids,
        "attention_mask": train_attention
    }, train_labels)).shuffle(1000).batch(16)

val_ds = tf.data.Dataset.from_tensor_slices(({
        "input_ids": val_input_ids,
        "attention_mask": val_attention
    }, val_labels)).batch(16)

device_name = "GPU" if tf.config.list_physical_devices("GPU") else "CPU"
print("Using device:", device_name)
```

```python
model = TFDistilBertForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=2)
```

```python

# =====================================
# 5. Compile Model
# =====================================
optimizer = tf.keras.optimizers.Adam(learning_rate=2e-5)
loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
metric = tf.keras.metrics.SparseCategoricalAccuracy("accuracy")

model.compile(optimizer=optimizer, loss=loss, metrics=[metric])

# =====================================
# 6. Train
# =====================================
EPOCHS = 3
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS
)

# =====================================
# 7. Evaluate
# =====================================
val_preds = model.predict(val_ds).logits
val_preds = np.argmax(val_preds, axis=1)
acc = accuracy_score(Y_test, val_preds)
print("Validation Accuracy:", acc)

```

```python
# =====================================
# 8. Save Model
# =====================================
model.save_pretrained("distilbert-tf-statement-pair")
tokenizer.save_pretrained("distilbert-tf-statement-pair")
```
